```python
#1) Import pandas, numpy

import pandas as pd
import numpy as np
#2) Load dataset using pd.read_csv()
url="https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv"
df=pd.read_csv(url)
df



```

```python


 #3) Display: shape, head, column names, datatypes, missing value
print(df.shape)

print(df.head())

print(df.columns)

print(df.dtypes)

print(df.isnull().sum())
```

```python


#4)Check and remove duplicates

print("Duplicate rows:")
print(df.duplicated().sum())

df = df.drop_duplicates()

print("Shape after removing duplicates:")
print(df.shape)
```

```python
#5)Rename columns ("Title" > "Movie_Title")

df.rename(columns={"Title": "Movie_Title"}, inplace=True)

print(df.columns)
```

```python
#6)Get unique genres

genres = sorted(df["Genre"].str.split(",").explode().unique())

print(genres)
```

```python
#7)Extract number of genres per movie

if "Movie_Title" not in df.columns and "Title" in df.columns:
    df.rename(columns={"Title": "Movie_Title"}, inplace=True)

df["Genre_Count"] = df["Genre"].apply(lambda x: len(x.split(",")))

print(df[["Movie_Title", "Genre", "Genre_Count"]].head())
```

```python
#8)Handle missing values in 'Revenue (Millions)' and 'Metascore

df["Revenue (Millions)"] = df["Revenue (Millions)"].fillna(
    df["Revenue (Millions)"].median()
)

df["Metascore"] = df["Metascore"].fillna(
    df["Metascore"].median()
)

print(df[["Revenue (Millions)", "Metascore"]].isnull().sum())
```

```python
#9)Top 5 movies by Rating

top_5 = df.sort_values(
    by="Rating",
    ascending=False
).head(5)

print(top_5[["Movie_Title", "Rating"]])
```

```python
#10)Movies with highest/lowest revenue
highest = df.loc[
    df["Revenue (Millions)"].idxmax()
]

lowest = df.loc[
    df["Revenue (Millions)"].idxmin()
]

print(highest[["Movie_Title", "Revenue (Millions)"]])
print(lowest[["Movie_Title", "Revenue (Millions)"]])
```

```python
#11) Avg. rating by Director
avg_rating = df.groupby("Director")["Rating"].mean()

print(avg_rating)
```

```python
#12)Movies per year
movies_per_year = df.groupby("Year")["Movie_Title"].count()

print(movies_per_year)
```

```python
#13)Movies with rating > 8 & revenue > 100M
result = df[
    (df["Rating"] > 8) &
    (df["Revenue (Millions)"] > 100)
]

print(
    result[
        ["Movie_Title", "Rating", "Revenue (Millions)"]
    ]
)
```

```python
#14)Group by Director: avg Rating, Revenue
director_group = df.groupby("Director").agg(
    Avg_Rating=("Rating", "mean"),
    Avg_Revenue=("Revenue (Millions)", "mean")
)

print(director_group)
```

```python
#15) Group by Year: count, avg Rating
year_group = df.groupby("Year").agg(
    Count=("Movie_Title", "count"),
    Avg_Rating=("Rating", "mean")
)

print(year_group)
```

```python
#16) Create a new column Profitability:
#a Revenue divided by Duration
df["Profitability"] = (
    df["Revenue (Millions)"] /
    df["Runtime (Minutes)"]
)

print(
    df[
        ["Movie_Title", "Revenue (Millions)",
         "Runtime (Minutes)", "Profitability"]
    ].head()
)
#b) Find top 10 most profitable movies
top_10 = df.sort_values(
    by="Profitability",
    ascending=False
).head(10)

print(
    top_10[
        ["Movie_Title", "Profitability"]
    ]
)
```

```python
#17) Create a column that categorizes movies by rating:
#a) Rating < 6: Poor
df["Rating_Category"] = np.where(df["Rating"] < 6, "Poor", "")

print(df[["Movie_Title", "Rating", "Rating_Category"]].head())
#b) 6 ≤ Rating < 7.5: Average
df["Rating_Category"] = np.where(
    (df["Rating"] >= 6) & (df["Rating"] < 7.5),
    "Average",
    ""
)

print(df[["Movie_Title", "Rating", "Rating_Category"]].head())
#c) Rating ≥ 7.5: Excellent
df["Rating_Category"] = np.where(
    df["Rating"] >= 7.5,
    "Excellent",
    ""
)

print(df[["Movie_Title", "Rating", "Rating_Category"]].head())
```

```python
#18) Create a custom function to score movies based on:
#i) Apply row['Rating'] * 10 + row['Revenue (Millions)'] as High score
#Apply a function to dataset to assign the score
def movie_score(row):
    return row["Rating"] * 10 + row["Revenue (Millions)"]

df["Score"] = df.apply(movie_score, axis=1)

print(df[["Movie_Title", "Rating", "Revenue (Millions)", "Score"]].head())
```
